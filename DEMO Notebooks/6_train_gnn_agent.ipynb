```python
from primaite.agents.gnn_agent import GNNAgent
```

```python
import os

laydowns = os.listdir("/srv/aegis-training-data/laydowns/group_1/laydowns/")
```

```python
agent = GNNAgent(
    training_config_path="../agents/training_configs/gnn.yaml",
    lay_down_config_path="/srv/aegis-evaluation-data/eval_dataset/5_nodes.yaml",
    gamma=0.999,
    actor_lr=0.005,
    critic_lr=0.005,
    hidden_dim=256,
    gat_layers=3,
    total_episodes=128,
)
```

```python
for laydown in laydowns:
    print(laydown)
    agent.reset_env("/srv/aegis-training-data/laydowns/group_1/laydowns/" + laydown)
    agent.learn(episodes=1)
```

```python
agent.reset_env(laydown_config_path="/srv/aegis-evaluation-data/eval_dataset/5_nodes.yaml")
```

```python
rewards = agent.evaluate(time_steps=128, episodes=3)
```

```python
agent.actor.save("./")
agent.critic.save("./")
```
