```python
from primaite.agents.gnn_agent import GNNAgent
```

```python
agent = GNNAgent(
    training_config_path="../agents/training_configs/gnn.yaml",
    lay_down_config_path="/srv/aegis-evaluation-data/eval_dataset/5_nodes.yaml",
    gamma=0.999,
    actor_lr=0.005,
    critic_lr=0.005,
    hidden_dim=256,
    gat_layers=3
)

rewards = agent.learn(
    time_steps=128,
    episodes=3
)
```

```python
rewards = agent.evaluate(
    time_steps=128,
    episodes=3
)
```
