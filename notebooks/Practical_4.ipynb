#### c. Task 1) Setup graph convolutional network for node classification
Define a GCN class for a node classification that takes five inputs: *input_dim*, *hid_dim*, *n_class*, *n_layers*, and *dropout_ratio*.

- Note that the `forward` function should return a Tensor object: **logits** and the `generate_node_embeddings` fuction should return a Tensor object: **node_embeddings**, which is the representation of the last layer.
- For the case when the *n_layers=0*, implement a single feed-forward network.
- Use `F.relu` and `F.dropout` at the end of each layer.


```python
class AttGNNLayer(nn.Module):

    def __init__(self, input_dim, output_dim, att_hid_dim=16, absolute=False):
        super(AttGNNLayer, self).__init__()
        self.W_self = nn.Linear(input_dim, output_dim)
        self.W_nbh = nn.Linear(input_dim, output_dim)
        self.bias = nn.Parameter(torch.zeros(output_dim))
        if absolute:
            self.attention = MLP([2 * input_dim, att_hid_dim, 1])
        else:
            self.attention = Sequential('x', [(lambda u: u[:, input_dim:] - u[:, :input_dim], 'x -> x'),
                                           (MLP([input_dim, att_hid_dim, 1]), 'x -> x')])


    def forward(self, node_feats, edge_index):
        # Compute attention for each pair of nodes

        e = self.attention(torch.cat([node_feats[edge_index[0, :]], node_feats[edge_index[1, :]]], 1))
        n = node_feats.size(dim=0)
        e_adj = -1e20*torch.ones(n, n).to(device)
        e_adj[edge_index[0, :], edge_index[1, :]] = torch.flatten(e)
        a_adj = F.softmax(e_adj, dim=1)

        # Aggregate
        return self.W_self(node_feats) + self.W_nbh(a_adj @ node_feats) + self.bias
```

```python
class GCN(nn.Module):
    def __init__(
        self,
        input_dim: int,
        hid_dim: int,
        n_classes: int,
        n_layers: int,
        dropout_ratio: float = 0.3):

        super(GCN, self).__init__()
        """
        Args:
        input_dim: input feature dimension
        hid_dim: hidden feature dimension
        n_classes: number of target classes
        n_layers: number of layers
        dropout_ratio: dropout_ratio
        """
        ## ------ Begin Solution ------ ##
        if n_layers >= 1:
            m = [(GATv2Conv(input_dim, hid_dim), 'x, edge_index -> x1'),
                 nn.ReLU(inplace=True),
                 (nn.Dropout(p=dropout_ratio), 'x1 -> x1')]
            for i in range(n_layers - 1):
                m += [(GATv2Conv(hid_dim, hid_dim), 'x1, edge_index -> x1'),
                      nn.ReLU(inplace=True),
                      (nn.Dropout(p=dropout_ratio), 'x1 -> x1')]
            self.net = Sequential('x, edge_index', m)
            rep_dim = hid_dim
        else:
            self.net = Sequential('x, edge_index',
                            [(nn.Identity(), 'x -> x1')])
            rep_dim = input_dim
        self.classifier = Sequential('x1',
                            [(nn.Linear(rep_dim, n_classes), 'x1 -> x2'),
                            nn.ReLU(inplace=True)])
        ## ------ End Solution ------ ##

    def forward(self, X, A) -> torch.Tensor:
        ## ------ Begin Solution ------ ##
        return self.classifier(self.net(X, A))
        ## ------ End Solution ------ ##

    def generate_node_embeddings(self, X, A) -> torch.Tensor:
        ## ------ Begin Solution ------ ##
        return self.net(X, A)
        ## ------ End Solution ------ ##

    def param_init(self):
        ## ------ Begin Solution ------ ##
        def init_weights(l):
            if isinstance(l, nn.Linear):
                nn.init.xavier_uniform_(l.weight.data)
                nn.init.normal_(l.bias.data, 0.0, 1.0)
            elif isinstance(l, torch_geometric.nn.GCNConv):
                l.reset_parameters()

        self.net.apply(init_weights)
        self.classifier.apply(init_weights)
        ## ------ End Solution ------ ##
```

#### c. Task 2) Implement `train` and `evaluate` functions for training a node classification model.

- Apply **Early-stopping**: stop training if the validation accuracy decreases during *k* consecutive epochs.
- Use **Adam optimizer** for training.
- Use **evaluate** function for calculating the validation accuracy in every epoch to adapt early-stopping.

Please note that `train` function should return the trained model object and `evaluate` function should return a validation accuracy.

```python
def accuracy(pred, target):
    return (pred == target).float().mean().item()
```

```python
def train(params: typing.Dict) -> torch.nn.Module:
    """
        This function trains a node classification model and returns the trained model object.
    """
    # set device
    device = "cuda" if torch.cuda.is_available() else "cpu"

    # load dataset
    data = dataset.data
    data = data.to(device)

    # Update parameters
    params["n_classes"] = dataset.num_classes # number of target classes
    params["input_dim"] = dataset.num_features # size of input features

    # Set a model
    if params['model_name'] == 'GCN':
        model = GCN(
            params["input_dim"],
            params["hid_dim"],
            params["n_classes"],
            params["n_layers"]
            ).to(device)
    elif params['model_name'] == 'SkipGCN':
        model = SkipGCN(
            params["input_dim"],
            params["hid_dim"],
            params["n_classes"],
            params["n_layers"]
        ).to(device)
    elif params['model_name'] == 'JumpKnowGCN':
        model = JumpKnowGCN(
            params["input_dim"],
            params["hid_dim"],
            params["n_classes"],
            params["n_layers"]
        ).to(device)
    else:
        raise NotImplementedError
    model.param_init()
    ## ------ Begin Solution ------ ##

    criterion = torch.nn.CrossEntropyLoss()

    optimizer = torch.optim.Adam(model.parameters(),
                                 lr=params["lr"],
                                 weight_decay=params["weight_decay"])
    mask = data.train_mask
    x = data.x
    target = data.y[mask]
    edge_index = data.edge_index

    plateau_count = 0
    max_acc = 0.0

    for epoch in range(params["epochs"]):
        if plateau_count > params["max_patience"]:
            print("Early stopped at epoch {}".format(epoch))
            break
        # set model to train mode
        model.train()

        # compute predicted logits and labels
        pred_logits = model(x, edge_index)[mask]
        pred = pred_logits.argmax(dim=1)

        loss = criterion(pred_logits, target)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        # for early stopping
        acc = evaluate(model, data, data.test_mask)
        if acc <= max_acc:
            plateau_count += 1
        else:
            max_acc = acc
            plateau_count = 0

    print("final accuracy: {}".format(max_acc))
    return model
    ## ------ End Solution ------ ##
```

```python
def evaluate(model,
             data,
             mask):
    ## ------ Begin Solution ------ ##

    model.eval()
    pred = model(data.x, data.edge_index).argmax(dim=1)[mask]
    target = data.y[mask]

    return accuracy(pred, target)
    ## ------ End Solution ------ ##
```

#### b. Task 3) Implement a function for demensionality reduction

- The `dimension_reduction` function will take a model object and data, perform T-SNE, and return a *DataFrame* that contains reduced variables and labels for each data point.
- Please reduce the dimension to 2d.
- Use validation data for the visualisation

** You may refer this address below using T-SNE:
[Link to sklearn T-SNE](https://scikit-learn.org/stable/modules/generated/sklearn.manifold.TSNE.html)

```python
def dimension_reduction(model: nn.Module) -> pd.DataFrame:
    """
        Args:
        model: model object for generating features

        Return:
        pd.DataFrame: A data frame that has 'dimension 1', 'dimension 2', and 'labels' as a column
    """
    ## ------ Begin Solution ------ ##
    data = dataset.data
    data.to("cpu")
    model.to("cpu")
    e = model.generate_node_embeddings(data.x, data.edge_index)[data.val_mask]
    e = e.detach().numpy()
    e_reduced = TSNE(n_components=2, learning_rate='auto',
                     init='random', perplexity=3).fit_transform(e)
    d = {'embedding_x': e_reduced[:, 0],
         'embedding_y': e_reduced[:, 1],
         'label': data.y[data.val_mask].detach().numpy()}

    p = pd.DataFrame(data=d)
    return p

    ## ------ End Solution ------ ##
```

## Part 2. Skip-Connections

In this section, we will investigate whether skip-connections can alleviate the effect of over-smoothing.

#### a. Task 5) Setup graph convolutional network with skip-connections
Define a SkipGCN class for node classification that takes five inputs: *input_dim*, *hid_dim*, *n_class*, *n_layers*, and *dropout_ratio*. Using **residual connection** is recommended, but you may also use other techniques, like **concatenation**.

- Note that the `forward` function should return a Tensor object: **logits** and the `generate_node_embeddings` fuction should return a Tensor object: **node_embeddings**, which is the representation of the last layer.
- For the case when the *n_layers=0*, implement a single feed-forward network.
- Use `F.relu` and `F.dropout` at the end of each layer.


```python
class SkipGCN(nn.Module):
    def __init__(self,
                 input_dim: int,
                 hid_dim: int,
                 n_classes: int,
                 n_layers: int,
                 dropout_ratio: float = 0.3):
        super(SkipGCN, self).__init__()
        """
        Args:
        input_dim: input feature dimension
        hid_dim: hidden feature dimension
        n_classes: number of target classes
        n_layers: number of layers
        dropout_ratio
        """

        ## ------ Begin Solution ------ ##
        if n_layers >= 1:
            m = [(GCNConv(input_dim, hid_dim), 'x, edge_index -> x'),
                  nn.ReLU(inplace=True),
                  (nn.Dropout(p=dropout_ratio), 'x -> x')]
            for i in range(n_layers - 1):
                m += [(GCNConv(hid_dim, hid_dim), 'x, edge_index -> nbh'),
                     (lambda a, b: a+b, 'x, nbh -> x'),
                     nn.ReLU(inplace=True),
                     (nn.Dropout(p=dropout_ratio), 'x -> x')]
            self.net = Sequential('x, edge_index', m)
            rep_dim = hid_dim
        else:
            self.net = Sequential('x, edge_index', [(nn.Identity(), 'x -> x')])
            rep_dim = input_dim

        self.classifier = Sequential('x',
                            [(nn.Linear(rep_dim, n_classes), 'x -> x')])
        ## ------ End Solution ------ ##

    def forward(self, X, A) -> torch.Tensor:
        ## ------ Begin Solution ------ ##
        return self.classifier(self.net(X, A))
        ## ------ End Solution ------ ##

    def generate_node_embeddings(self, X, A) -> torch.Tensor:
        ## ------ Begin Solution ------ ##
        return self.net(X, A)
        ## ------ End Solution ------ ##

    def param_init(self):
        ## ------ Begin Solution ------ ##
        def init_weights(l):
            if isinstance(l, nn.Linear):
                nn.init.xavier_uniform_(l.weight.data)
                nn.init.normal_(l.bias.data, 0.0, 1.0)
            elif isinstance(l, torch_geometric.nn.GCNConv):
                l.reset_parameters()

        self.net.apply(init_weights)
        self.classifier.apply(init_weights)
        ## ------ End Solution ------ ##
```

## Optional) Jumping Knowledge Connection

#### a. Task6) Design Graph Convolutional Network with Jumping Knowledge Connection
Define a `JumpKnowGCN` class for a node classification that takes five inputs: *input_dim*, *hid_dim*, *n_class*, *n_layers*, and *dropout_ratio*. Using `max_pooling` is recommended but you may also use other differentiable functions.

- Note that the `forward` function should return a Tensor object: **logits** and the `generate_node_embeddings` fuction should return a Tensor object: **node_embeddings**, which is the representation of the last layer.
- For the case when the *n_layers=0*, implement a single feed-forward network.
- Use `relu` and `dropout` at the end of each layer.


```python
from torch_geometric.nn import JumpingKnowledge

class JumpKnowGCN(nn.Module):
    def __init__(
      self,
      input_dim: int,
      hid_dim: int,
      n_classes: int,
      n_layers: int,
      dropout_ratio: float = 0.3):
        super(JumpKnowGCN, self).__init__()
        """
        Args:
        input_dim: input feature dimension
        hid_dim: hidden feature dimension
        n_classes: number of target classes
        n_layers: number of layers
        dropout_ratio
        """

        ## ------ Begin Solution ------ ##
        if n_layers >= 1:
            m = [(AttGNNLayer(input_dim, hid_dim), 'x, edge_index -> x'),
                  nn.ReLU(inplace=True),
                  (nn.Dropout(p=dropout_ratio), 'x -> x'),
                  (lambda u: [u], 'x -> xs')
                  ]
            for i in range(n_layers - 1):
                m += [(GATv2Conv(hid_dim, hid_dim), 'x, edge_index -> x'),
                     nn.ReLU(inplace=True),
                     (nn.Dropout(p=dropout_ratio), 'x -> x'),
                     (lambda u, us: us + [u], 'x, xs -> xs')]
            m += [(JumpingKnowledge("max"), 'xs -> x')]
            self.net = Sequential('x, edge_index', m)
            rep_dim = hid_dim
        else:
            self.net = Sequential('x, edge_index', [(nn.Identity(), 'x -> x')])
            rep_dim = input_dim

        self.classifier = Sequential('x',
                            [(nn.Linear(rep_dim, n_classes), 'x -> x')])
        ## ------ End Solution ------ ##

    def forward(self, X, A) -> torch.Tensor:
        ## ------ Begin Solution ------ ##
        return self.classifier(self.net(X, A))
        ## ------ End Solution ------ ##

    def generate_node_embeddings(self, X, A) -> torch.Tensor:
        ## ------ Begin Solution ------ ##
        return self.net(X, A)
        ## ------ End Solution ------ ##

    def param_init(self):
        ## ------ Begin Solution ------ ##
        def init_weights(l):
            if isinstance(l, nn.Linear):
                nn.init.xavier_uniform_(l.weight.data)
                nn.init.normal_(l.bias.data, 0.0, 1.0)
            elif isinstance(l, torch_geometric.nn.GCNConv):
                l.reset_parameters()

        self.net.apply(init_weights)
        self.classifier.apply(init_weights)
        ## ------ End Solution ------ ##
```
